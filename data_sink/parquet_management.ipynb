# Parquet File Management

This notebook provides tools to inspect and manage parquet files in the data lake directory.

```python
import os
import glob
import pandas as pd
from pathlib import Path

# Get the data lake directory path
DATA_DIR = os.path.join(os.path.abspath(''), "data_lake")
```

## 1. Inspect Parquet Files

This section allows you to view the contents of parquet files and analyze the data.

```python
def inspect_parquet_files():
    """
    Inspects all parquet files in the data lake directory.
    Displays the combined data and product statistics.
    """
    # Get list of parquet files
    parquet_files = glob.glob(os.path.join(DATA_DIR, "*.parquet"))
    
    if not parquet_files:
        print("No parquet files found in data lake directory")
        print("No data available to analyze")
        return None
    
    # Read and combine all parquet files
    df = pd.concat([pd.read_parquet(f) for f in parquet_files])
    
    # Display the raw data
    print("\nRaw Data:")
    display(df)
    
    # Calculate and display product statistics
    product_counts = df.groupby('product').size().reset_index(name='count')
    product_counts = product_counts.sort_values(by='count', ascending=False)
    
    print("\nProduct Statistics:")
    display(product_counts)
    
    return df

# Run the inspection
df = inspect_parquet_files()
```

## 2. Clear Parquet Files

This section allows you to clear all parquet files from the data lake directory.

```python
def clear_parquet_files():
    """
    Removes all parquet files from the data lake directory.
    Prints the number of files deleted and their names.
    """
    # Get all parquet files in the data lake directory
    parquet_files = glob.glob(os.path.join(DATA_DIR, "*.parquet"))
    
    if not parquet_files:
        print("No parquet files found in the data lake directory.")
        return
    
    # Delete each parquet file
    for file_path in parquet_files:
        try:
            os.remove(file_path)
            print(f"Deleted: {os.path.basename(file_path)}")
        except Exception as e:
            print(f"Error deleting {file_path}: {str(e)}")
    
    print(f"\nSuccessfully deleted {len(parquet_files)} parquet files.")

# Uncomment the line below to clear parquet files
# clear_parquet_files()
```

This section provides additional analysis options if data is available.
